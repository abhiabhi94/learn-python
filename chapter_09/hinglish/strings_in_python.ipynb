# Strings in Python

Pehle chapters me humne strings ko thoda bahut use kiya. Ab hum strings ke baare me detail me seekhenge.

Strings text data ko represent karti hai - jaise names, addresses, messages wagera. Python me strings bahut powerful hai aur inke saath kai operations kar sakte hai.

Socho strings ek mala (garland) hai jisme har phool ek character hai. Hum is mala ko tod sakte hai, jod sakte hai, aur kai tarah se use kar sakte hai!

## String kya hai?

String characters ka sequence hai. Characters matlab letters, numbers, symbols - sab kuch!

Python me strings ko quotes me likhte hai:
- Single quotes: `'hello'`
- Double quotes: `"hello"`
- Triple quotes: `'''hello'''` ya `"""hello"""`

````python
# Different ways to create strings
name = 'Ram'
city = "Delhi"
message = '''Yeh ek
multi-line
string hai'''

print(name)
print(city)
print(message)
````

## String Operations

Strings ke saath hum kai operations kar sakte hai.

### Concatenation (Jodna)

````python
first_name = 'Amit'
last_name = 'Sharma'

# + se jod sakte hai
full_name = first_name + ' ' + last_name
print(full_name)
````

### Repetition (Repeat karna)

````python
laugh = 'ha'
print(laugh * 5)  # hahahahahaha
````

### Length (Lambai)

````python
message = 'Namaste Duniya'
print('Length:', len(message))
````

## String Indexing

Har character ki position ko index kehte hai. Python me indexing 0 se start hoti hai.

```
String: H  E  L  L  O
Index:  0  1  2  3  4
```

````python
word = 'HELLO'

print('Pehla character:', word[0])
print('Doosra character:', word[1])
print('Last character:', word[4])

# Negative indexing - end se count karo
print('Last character:', word[-1])
print('Second last:', word[-2])
````

## String Slicing

String ka ek part nikalne ko slicing kehte hai.

**Syntax:** `string[start:end:step]`

````python
text = 'Python Programming'

# Pehle 6 characters
print(text[0:6])  # Python
print(text[:6])   # Same

# 7th se end tak
print(text[7:])   # Programming

# Har doosra character
print(text[::2])  # Pto rgamn

# Reverse
print(text[::-1]) # gnimmargorP nohtyP
````

## String Methods

Python me strings ke liye kai ready-made methods hai.

### Case conversion

````python
text = 'Hello World'

print(text.upper())      # HELLO WORLD
print(text.lower())      # hello world
print(text.title())      # Hello World
print(text.capitalize()) # Hello world
````

### Whitespace removal

````python
text = '   hello   '

print(text.strip())   # 'hello'
print(text.lstrip())  # 'hello   '
print(text.rstrip())  # '   hello'
````

### Find aur Replace

````python
sentence = 'I love Python programming'

# Find - position batata hai
print(sentence.find('Python'))  # 7
print(sentence.find('Java'))    # -1 (not found)

# Replace
new_sentence = sentence.replace('Python', 'Java')
print(new_sentence)
````

### Split aur Join

````python
# Split - string ko list me convert karta hai
sentence = 'Aam kela seb angoor'
fruits = sentence.split(' ')
print(fruits)

# Join - list ko string me convert karta hai
new_sentence = ' aur '.join(fruits)
print(new_sentence)
````

### Checking methods

````python
text = 'Hello123'

print(text.isalpha())     # False (numbers hai)
print(text.isdigit())     # False (letters hai)
print(text.isalnum())     # True (letters + numbers)
print('123'.isdigit())    # True
print('Hello'.isalpha())  # True
````

````python
# Aur bhi methods
print('hello'.startswith('he'))  # True
print('hello'.endswith('lo'))    # True
print('hello world'.count('l'))  # 3
````

## String Formatting

Variables ko string me include karne ke tarike:

### 1. % formatting (purana tarika)

````python
name = 'Ram'
age = 25

message = 'Mera naam %s hai aur main %d saal ka hoon' % (name, age)
print(message)
````

### 2. .format() method

````python
message = 'Mera naam {} hai aur main {} saal ka hoon'.format(name, age)
print(message)

# Position specify kar sakte hai
message = '{1} saal ka {0}'.format(name, age)
print(message)
````

### 3. f-strings (Python 3.6+, sabse best!)

````python
name = 'Priya'
age = 22
city = 'Mumbai'

# f-string use karke
message = f'Mera naam {name} hai, main {age} saal ki hoon aur {city} se hoon'
print(message)

# Expressions bhi likh sakte hai
print(f'Next year main {age + 1} saal ki houngi')
````

## String ke saath looping

````python
word = 'PYTHON'

# Har character par loop
for char in word:
    print(char)

print()

# Index ke saath
for i, char in enumerate(word):
    print(f'Index {i}: {char}')
````

## Escape Characters

Special characters ko represent karne ke liye backslash `\` use karte hai.

````python
# New line
print('Line 1\nLine 2')

# Tab
print('Name:\tRam')

# Quote string ke andar
print('It\'s a beautiful day')

# Backslash
print('Path: C:\\Users\\Ram')
````

## Raw Strings

Agar escape characters ko ignore karna ho, to raw string use karo (r se start).

````python
# Normal string
path = 'C:\\Users\\Ram\\Documents'
print(path)

# Raw string
path = r'C:\Users\Ram\Documents'
print(path)
````

## String Immutability

Python me strings **immutable** hai - matlab ek baar ban gaya to change nahi kar sakte.

Par nayi string bana sakte hai!

````python
text = 'Hello'

# Ye error dega
# text[0] = 'h'

# Par ye kar sakte hai
text = 'h' + text[1:]
print(text)  # hello
````
