## Class kya hai?

Class ek blueprint/template hai. Jaise ghar banane ka naksha.

**Object** class ka instance hai - matlab blueprint se bani hui cheez.

### Pehli class banate hai

```python
class Person:
    # Class variable
    species = 'Human'
    
    # Constructor - object banate waqt ye method call hota hai
    def __init__(self, name, age):
        # Instance variables
        self.name = name
        self.age = age
    
    # Method
    def introduce(self):
        print(f'Namaste! Mera naam {self.name} hai aur main {self.age} saal ka hoon.')
```

### Object banate hai

```python
# Object create karna
person1 = Person('Ram', 25)
person2 = Person('Priya', 22)

# Method call karna
person1.introduce()
person2.introduce()

# Attributes access karna
print(f'\n{person1.name} ki species: {person1.species}')
```

**Important points:**

1. `__init__` constructor hai - object banate waqt ye automatically call hota hai
2. `self` current object ko refer karta hai
3. Instance variables har object ke liye alag hote hai
4. Class variables sabhi objects ke liye same hote hai

## Encapsulation

Data aur methods ko ek saath rakhna aur kuch cheezein private rakhna.

```python
class BankAccount:
    def __init__(self, account_holder, balance=0):
        self.account_holder = account_holder
        self.__balance = balance  # Private variable (__ se start)
    
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f'{amount} rupaye jama ho gaye. Balance: {self.__balance}')
    
    def withdraw(self, amount):
        if amount > self.__balance:
            print('Balance kam hai!')
        else:
            self.__balance -= amount
            print(f'{amount} rupaye nikale. Balance: {self.__balance}')
    
    def get_balance(self):
        return self.__balance
```

```python
account = BankAccount('Ram', 1000)
account.deposit(500)
account.withdraw(300)
print(f'Final balance: {account.get_balance()}')
```

## Inheritance (Virasat)

Jaise baap se beta ko properties milti hai, waise ek class doosri class se properties le sakti hai.

```python
# Parent class
class Animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        print(f'{self.name} kuch awaaz kar raha hai')

# Child class
class Dog(Animal):
    def speak(self):
        print(f'{self.name} bhaunk raha hai: Bhow Bhow!')

class Cat(Animal):
    def speak(self):
        print(f'{self.name} mew kar rahi hai: Meow!')
```

```python
dog = Dog('Tommy')
cat = Cat('Milly')

dog.speak()
cat.speak()
```

## Polymorphism

Same method, different behavior - alag alag objects me alag tarike se kaam karna.

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    def area(self):
        return self.width * self.height

class Circle:
    def __init__(self, radius):
        self.radius = radius
    
    def area(self):
        return 3.14 * self.radius ** 2

# Dono me area() method hai par alag tarike se kaam karta hai
rect = Rectangle(5, 3)
circle = Circle(7)

print(f'Rectangle area: {rect.area()}')
print(f'Circle area: {circle.area()}')
```

## Special Methods (Magic Methods)

Python me kuch special methods hai jo __ se start aur end hote hai.

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
    
    # String representation
    def __str__(self):
        return f'"{self.title}" by {self.author}'
    
    # Length
    def __len__(self):
        return self.pages
    
    # Comparison
    def __gt__(self, other):
        return self.pages > other.pages
```

```python
book1 = Book('Gitanjali', 'Tagore', 200)
book2 = Book('Godan', 'Premchand', 300)

print(book1)  # __str__ call hoga
print(f'Pages: {len(book1)}')  # __len__ call hoga
print(f'Book2 badi hai? {book2 > book1}')  # __gt__ call hoga
```
