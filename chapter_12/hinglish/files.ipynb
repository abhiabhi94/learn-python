# Playing with Files

Computer me data ko permanently store karne ke liye files ka upyog hota hai. Python se hum files ko read, write, aur modify kar sakte hai.

Jaise aap notebook me notes likhte ho aur baad me padhte ho, waise hi programs bhi files me data likh sakte hai aur baad me padh sakte hai!

## File kholna (Opening)

`open()` function se file ko kholte hai.

**Syntax:** `open(filename, mode)`

**Modes:**
- `'r'` - Read (default)
- `'w'` - Write (overwrite karta hai)
- `'a'` - Append (end me add karta hai)
- `'r+'` - Read aur Write
- `'x'` - Create (already exist kare to error)

## File padhna (Reading)

```python
# File banate hai pehle
with open('sample.txt', 'w') as f:
    f.write('Namaste Duniya!\n')
    f.write('Python seekh rahe hai\n')
    f.write('Files bahut useful hai\n')
```

### Poori file ek saath padhna

```python
# read() method
with open('sample.txt', 'r') as f:
    content = f.read()
    print(content)
```

### Line by line padhna

```python
# readline() - ek line padhta hai
with open('sample.txt', 'r') as f:
    line1 = f.readline()
    line2 = f.readline()
    print('Pehli line:', line1)
    print('Doosri line:', line2)
```

```python
# readlines() - sab lines list me
with open('sample.txt', 'r') as f:
    lines = f.readlines()
    print(lines)
```

```python
# Loop se padhna (best way!)
with open('sample.txt', 'r') as f:
    for line in f:
        print(line.strip())  # strip() newline character hata deta hai
```

## File me likhna (Writing)

### 'w' mode - overwrite karta hai

```python
with open('output.txt', 'w') as f:
    f.write('Pehli line\n')
    f.write('Doosri line\n')
```

### 'a' mode - append karta hai

```python
with open('output.txt', 'a') as f:
    f.write('Teesri line\n')
    f.write('Chauthi line\n')

# Dekh lete hai
with open('output.txt', 'r') as f:
    print(f.read())
```

## with statement

`with` statement ka upyog karke file automatically close ho jati hai. Ye best practice hai!

```python
# Bina with ke
f = open('test.txt', 'w')
f.write('Hello')
f.close()  # Manually close karna padta hai

# with ke saath (better!)
with open('test.txt', 'w') as f:
    f.write('Hello')  # Automatically close ho jayegi
```

## File Operations

```python
import os

# File exist karti hai?
print('Exists?', os.path.exists('sample.txt'))

# File delete karna
if os.path.exists('test.txt'):
    os.remove('test.txt')
    print('File deleted')
```

## CSV Files

CSV (Comma Separated Values) files data store karne ke liye bahut common hai.

```python
import csv

# CSV file likhna
data = [
    ['Naam', 'Umar', 'Shehar'],
    ['Ram', '25', 'Delhi'],
    ['Priya', '22', 'Mumbai'],
    ['Amit', '30', 'Bangalore']
]

with open('data.csv', 'w', newline='') as f:
    writer = csv.writer(f)
    writer.writerows(data)
```

```python
# CSV file padhna
with open('data.csv', 'r') as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)
```

## JSON Files

JSON (JavaScript Object Notation) data exchange ke liye popular hai.

```python
import json

# Dictionary ko JSON me convert
data = {
    'name': 'Ram',
    'age': 25,
    'city': 'Delhi',
    'hobbies': ['reading', 'coding']
}

# JSON file me likhna
with open('data.json', 'w') as f:
    json.dump(data, f, indent=2)
```

```python
# JSON file padhna
with open('data.json', 'r') as f:
    loaded_data = json.load(f)
    print(loaded_data)
    print(type(loaded_data))
```
