## If-Else Statements

If-else ka matlab hai "agar-warna". Ye humko decisions lene me madad karta hai.

Socho ki aap kisi restaurant me khana khane gaye ho. Menu me daal hai ya nahi, ye check karna chahte ho:
- **Agar** daal hai, to order karo
- **Warna** kuch aur order karo

Programming me bhi kuch aisa hi hota hai!

### Ek simple example

Chaliye dekhte hai ki hum kaise check kar sakte hai ki ek number even hai ya odd.

````python
num = int(input('Ek number enter karo: '))
````

Pehle hum check karte hai ki number ko 2 se divide karne par remainder kya aata hai:

````python
num % 2 == 0
````

Ab if-else ka upyog karke dekhte hai:

````python
num = int(input('Ek number enter karo: '))
if num % 2 == 0:
    print('Number even hai')
    print('Number ka square:', num ** 2)
else:
    print('Number odd hai')
    print('Number ka cube:', num ** 3)
````

**Dhyan dene wali baatein:**

1. `if` ke baad condition likhte hai aur uske baad colon `:` lagana zaruri hai
2. If ke andar wala code **indented** (aage se shuru) hona chahiye
3. `else:` ke baad bhi colon lagta hai
4. Python me indentation (spacing) bahut important hai!

### Nested If-Else

Kabhi kabhi humko ek decision ke andar doosra decision lena padta hai. Jaise:
- Agar paisa hai, to market jao
  - Agar market me sabzi hai, to kharido
  - Warna ghar wapas aao

Isko nested if-else kehte hai.

Teen numbers me se sabse bada number dhoondhne ka example:

````python
num_1 = int(input('Pehla number enter karo: '))
num_2 = int(input('Doosra number enter karo: '))
num_3 = int(input('Teesra number enter karo: '))

if num_1 > num_2:
    if num_1 > num_3:
        print(num_1, 'sabse bada hai')
    else:
        print(num_3, 'sabse bada hai')
else:
    if num_2 > num_3:
        print(num_2, 'sabse bada hai')
    else:
        print(num_3, 'sabse bada hai')
````

## Elif

`elif` ka matlab hai "else if" - yaani "warna agar".

Jab humko kai conditions check karni ho, to har baar nested if likhne se code bahut complicated ho jata hai. Uske liye hum `elif` ka upyog karte hai.

Ye excessive indentation se bachne me madad karta hai aur code ko readable banata hai.

### Purane code ko elif se improve karte hai

Pehle humne teen numbers me sabse bada dhoondhne ke liye nested if use kiya tha. Ab elif se karte hai:

````python
num_1 = int(input('Pehla number: '))
num_2 = int(input('Doosra number: '))
num_3 = int(input('Teesra number: '))

if num_1 > num_2 and num_1 > num_3:
    print(num_1, 'sabse bada hai')
elif num_2 > num_3:
    print(num_2, 'sabse bada hai')
else:
    print(num_3, 'sabse bada hai')
````

Dekho kitna saaf aur simple lag raha hai!

## Loops (Chakkar)

Kabhi kabhi humko koi kaam baar baar karna padta hai. Jaise:
- Roz subah brush karna
- Ek list me har item ko check karna
- 1 se 100 tak numbers print karna

Iske liye hum loops ka upyog karte hai. Python me do tarah ke loops hai:
1. **for loop** - jab pata ho kitni baar chalana hai
2. **while loop** - jab kisi condition tak chalana ho

## For Loop

`for` loop ka upyog tab karte hai jab humko pata ho ki kitni baar loop chalana hai, ya jab hume kisi sequence (list, string, etc.) ke har element par kuch karna ho.

### Basic Syntax

```python
for item in sequence:
    # yaha code likho
```

### Ek simple example

````python
fruits = ['aam', 'kela', 'angoor', 'seb']

for fruit in fruits:
    print('Mujhe', fruit, 'pasand hai')
````

## Range Function

`range()` function ka upyog numbers ki sequence banane ke liye hota hai. Ye for loop ke saath bahut useful hai.

````python
# 0 se 4 tak (5 numbers)
for i in range(5):
    print(i)
````

````python
# 1 se 10 tak
for i in range(1, 11):
    print(i)
````

````python
# 2-2 karke badhna (even numbers)
for i in range(0, 11, 2):
    print(i)
````

### Abhyaas: Table banana

User se ek number lo aur uska table pehle 10 values tak print karo.

````python
num = int(input('Kaunsa table chahiye? '))

for i in range(1, 11):
    print(f'{num} x {i} = {num * i}')
````

### Abhyaas

Pehle 10 even natural numbers ka sum aur product nikalo.

## Nested Loops

Loop ke andar loop - jaise building me floors aur har floor me rooms!

Patterns banane ke liye ye bahut useful hai.

### Pattern printing

Chaliye ek pattern print karte hai:
```
*
**
***
****
*****
```

````python
for i in range(1, 6):
    for j in range(i):
        print('*', end='')
    print()  # Naye line ke liye
````

## While Loop

`while` loop tab tak chalta rehta hai jab tak ek condition True hai.

Jaise: "Jab tak bhook lagi hai, tab tak khana khao"

### Basic Syntax

```python
while condition:
    # code yaha
```

### Example: Pehle 10 natural numbers print karo

````python
i = 1
while i <= 10:
    print(i)
    i += 1  # i ko 1 se badhao
````

**Dhyan do:** While loop me counter ko khud badhana padta hai, warna loop infinite times chal sakta hai!

## Break Statement

`break` ka upyog loop ko beech me hi rokne ke liye hota hai.

Jaise: "Jab tak samosa nahi mil jata, tabtak dhoondhte raho. Mil gaya to ruk jao!"

### Example

````python
# 1 se 100 tak numbers print karo, par 50 pe ruk jao
for i in range(1, 101):
    if i == 50:
        break
    print(i)
````

### Abhyaas: Prime number check

User se ek number lo aur batao ki wo prime hai ya nahi.

````python
num = int(input('Number enter karo: '))

if num < 2:
    print('Prime nahi hai')
else:
    is_prime = True
    for i in range(2, num):
        if num % i == 0:
            is_prime = False
            break
    
    if is_prime:
        print('Prime hai')
    else:
        print('Prime nahi hai')
````

## Continue Statement

`continue` loop ke current iteration ko skip kar deta hai aur next iteration par chala jata hai.

Jaise: "Agar vegetable kharab hai, to usko chodo aur agle par jao."

### Example: Sirf odd numbers print karo

````python
for i in range(1, 11):
    if i % 2 == 0:  # Agar even hai
        continue    # to skip karo
    print(i)  # Sirf odd numbers print honge
````

## Looping Techniques

Python me kuch special techniques hai loops ko aur bhi powerful banane ke liye.

### Dictionary ke through loop

````python
fruits_price = {'aam': 80, 'kela': 40, 'seb': 120}

# Keys aur values dono chahiye
for fruit, price in fruits_price.items():
    print(f'{fruit} ki price hai {price} rupaye')
````

### Enumerate - index ke saath elements

````python
fruits = ['aam', 'kela', 'angoor']

for index, fruit in enumerate(fruits):
    print(f'{index}: {fruit}')
````

### Zip - Do lists ko saath me loop karna

````python
names = ['Ram', 'Shyam', 'Mohan']
ages = [25, 30, 22]

for name, age in zip(names, ages):
    print(f'{name} ki umar {age} saal hai')
````

### Reverse order me loop

````python
for i in reversed(range(1, 6)):
    print(i)
````

### Sorted order me loop

````python
numbers = [5, 2, 8, 1, 9]

for num in sorted(numbers):
    print(num)
````

## Exceptions (Galtiyon ko handle karna)

Kabhi kabhi program chalate waqt error aa jati hai. Jaise:
- Number ki jagah string daal diya
- Zero se divide karne ki koshish ki
- File nahi mili

Exception handling se hum in errors ko gracefully handle kar sakte hai, taaki program crash na ho.

### Bina exception handling

````python
# Ye error dega agar user number nahi dalega
num = int(input('Number dalo: '))
````

### Exception handling ke saath

````python
try:
    num = int(input('Number dalo: '))
    result = 100 / num
    print('Result:', result)
except ValueError:
    print('Galat input! Number dalo.')
except ZeroDivisionError:
    print('Zero se divide nahi kar sakte!')
````

**Try-Except ka structure:**

- `try:` block me wo code likho jo error de sakta hai
- `except:` block me batao ki error aane par kya karna hai

### Finally block

`finally` block hamesha execute hota hai, chahe error aaye ya na aaye.

````python
try:
    num = int(input('Number: '))
    print(100 / num)
except:
    print('Kuch galat ho gaya')
finally:
    print('Ye hamesha print hoga')
````
