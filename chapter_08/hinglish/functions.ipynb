## Functions kya hai?

Function ek code ka block hai jo:
- Ek specific kaam karta hai
- Naam se pehchana jata hai
- Jab zarurat ho tab call/use kiya ja sakta hai
- Optionally, kuch input le sakta hai aur kuch output de sakta hai

Functions se code:
- **Organized** rehta hai
- **Reusable** ban jata hai
- **Readable** ho jata hai

## Function kaise banaye?

Python me function banane ke liye `def` keyword ka upyog karte hai.

**Basic Syntax:**
```python
def function_name():
    # code yaha likho
```

### Ek simple function

````python
def greet():
    print('Namaste!')
    print('Kaise ho aap?')
````

Function define ho gaya, ab isko use karte hai:

````python
greet()  # Function ko call karna
````

Dekha! Jitni baar chahiye utni baar call kar sakte hai:

````python
greet()
greet()
greet()
````

## Parameters (Inputs)

Kabhi kabhi function ko kuch information chahiye hoti hai. Jaise chai banate waqt kitni cheeni dalein?

Is information ko **parameters** ya **arguments** kehte hai.

### Function with parameters

````python
def greet_person(name):
    print(f'Namaste {name}!')
    print('Kaise ho?')
````

````python
greet_person('Ram')
greet_person('Shyam')
````

### Multiple parameters

````python
def introduce(name, age, city):
    print(f'Mera naam {name} hai')
    print(f'Main {age} saal ka hoon')
    print(f'Main {city} se hoon')
````

````python
introduce('Amit', 25, 'Delhi')
introduce('Priya', 22, 'Mumbai')
````

## Default Parameters

Kabhi kabhi hum parameter ki ek default value rakhna chahte hai. Agar user kuch nahi deta, to default value use ho jaye.

````python
def make_chai(cups=1, sugar_spoons=2):
    print(f'{cups} cup chai ban rahi hai')
    print(f'{sugar_spoons} spoon cheeni ke saath')
````

````python
make_chai()  # Default values use hongi
make_chai(2)  # 2 cups, default sugar
make_chai(3, 1)  # 3 cups, 1 spoon sugar
````

### Keyword arguments

Hum parameter ka naam specify karke bhi call kar sakte hai. Isse order matter nahi karta.

````python
make_chai(sugar_spoons=3, cups=2)
````

## Return Statement

Function kuch calculate karke result de sakta hai. Iske liye `return` keyword ka upyog hota hai.

Jaise calculator - aap numbers dete ho, wo result deta hai.

### Simple example

````python
def add(a, b):
    result = a + b
    return result
````

````python
sum_value = add(5, 3)
print('Sum:', sum_value)
````

````python
# Direct bhi use kar sakte hai
print('Sum:', add(10, 20))
````

### Multiple values return karna

````python
def get_name_and_age():
    name = 'Rahul'
    age = 30
    return name, age
````

````python
person_name, person_age = get_name_and_age()
print(f'{person_name} ki umar {person_age} hai')
````

### Bina return ke function

Agar function me return nahi hai, to automatically `None` return hota hai.

````python
def say_hello():
    print('Hello!')

result = say_hello()
print('Result:', result)  # None milega
````

## Variable Scope (Reach)

Variables ka scope matlab hai ki wo kaha-kaha use ho sakte hai.

Do tarah ke variables hote hai:
1. **Local variables**: Function ke andar define, sirf wahi use ho sakte hai
2. **Global variables**: Function ke bahar define, kahin bhi use ho sakte hai

````python
# Global variable
city = 'Delhi'

def print_city():
    # Function ke andar se global variable access kar sakte hai
    print('City:', city)

print_city()
````

````python
def test_local():
    # Local variable
    message = 'Ye local hai'
    print(message)

test_local()
# print(message)  # Ye error dega, kyunki message local hai
````

## Lambda Functions

Lambda ek chhota, anonymous (bina naam ka) function hai. Jab koi simple operation ek baar karna ho, to lambda convenient hai.

**Syntax:**
```python
lambda arguments: expression
```

````python
# Normal function
def square(x):
    return x * x

# Same cheez lambda se
square_lambda = lambda x: x * x

print(square(5))
print(square_lambda(5))
````

````python
# Multiple parameters
add = lambda a, b: a + b
print(add(3, 4))
````

Lambda functions list operations me bahut useful hai:

````python
numbers = [1, 2, 3, 4, 5]

# Har number ka square
squares = list(map(lambda x: x**2, numbers))
print(squares)
````

## *args aur **kwargs

Kabhi kabhi pata nahi hota ki kitne arguments aayenge. Tab ye use karte hai:
- `*args`: Variable number of positional arguments
- `**kwargs`: Variable number of keyword arguments

### *args example

````python
def add_all(*numbers):
    total = 0
    for num in numbers:
        total += num
    return total

print(add_all(1, 2, 3))
print(add_all(1, 2, 3, 4, 5))
print(add_all(10, 20))
````

### **kwargs example

````python
def print_info(**info):
    for key, value in info.items():
        print(f'{key}: {value}')

print_info(name='Ram', age=25, city='Delhi')
print()
print_info(fruit='Aam', color='Yellow', price=80)
````

## Docstrings

Function ko document karne ke liye docstring use karte hai. Ye function ke baare me information deta hai.

Triple quotes `"""` ya `'''` me likhte hai.

````python
def calculate_area(length, width):
    """
    Rectangle ka area calculate karta hai.
    
    Parameters:
        length: Rectangle ki lambai
        width: Rectangle ki chaudai
    
    Returns:
        Area ki value
    """
    return length * width
````

````python
# Docstring ko dekh sakte hai
help(calculate_area)
````

## Recursion

Jab ek function khud ko call kare, usko recursion kehte hai.

Jaise mirrors ke samne mirrors - infinite reflections!

**Dhyan rakho:** Recursion me base condition zaruri hai, nahi to infinite loop ban jayega.

### Factorial example

5! = 5 × 4 × 3 × 2 × 1 = 120

````python
def factorial(n):
    # Base case
    if n == 0 or n == 1:
        return 1
    # Recursive case
    else:
        return n * factorial(n - 1)

print('5! =', factorial(5))
print('7! =', factorial(7))
````

## Python ke Built-in Functions

Python me pehle se hi kai useful functions hai. Humne kuch already use bhi kiye hai:

````python
# len() - length nikalta hai
fruits = ['aam', 'kela', 'seb']
print('Total fruits:', len(fruits))

# max() - maximum value
numbers = [10, 5, 20, 15]
print('Maximum:', max(numbers))

# min() - minimum value
print('Minimum:', min(numbers))

# sum() - sum nikalta hai
print('Sum:', sum(numbers))

# type() - data type batata hai
print('Type:', type(fruits))
````
