## Thread kya hai?

Thread ek chota program hai jo main program ke sath chalta hai. Multiple threads ek sath chal sakte hai.

```python
import threading
import time
```

## Simple Thread

```python
def print_numbers():
    for i in range(5):
        print(f'Number: {i}')
        time.sleep(1)

def print_letters():
    for letter in 'ABCDE':
        print(f'Letter: {letter}')
        time.sleep(1)

# Threads banao
t1 = threading.Thread(target=print_numbers)
t2 = threading.Thread(target=print_letters)

# Start karo
t1.start()
t2.start()

# Wait karo done hone tak
t1.join()
t2.join()

print('Done!')
```

## Arguments pass karna

```python
def greet(name, times):
    for i in range(times):
        print(f'Namaste {name}!')
        time.sleep(0.5)

t1 = threading.Thread(target=greet, args=('Ram', 3))
t2 = threading.Thread(target=greet, args=('Priya', 3))

t1.start()
t2.start()

t1.join()
t2.join()
```

## Thread Class inherit karna

```python
class MyThread(threading.Thread):
    def __init__(self, name):
        threading.Thread.__init__(self)
        self.name = name
    
    def run(self):
        for i in range(3):
            print(f'{self.name}: {i}')
            time.sleep(0.5)

t1 = MyThread('Thread-1')
t2 = MyThread('Thread-2')

t1.start()
t2.start()

t1.join()
t2.join()
```

## Lock - Race Condition se bachna

Jab multiple threads ek hi variable ko modify karte hai, to problem ho sakti hai. Lock se ye solve hota hai.

```python
counter = 0
lock = threading.Lock()

def increment():
    global counter
    for _ in range(100000):
        lock.acquire()
        counter += 1
        lock.release()

threads = [threading.Thread(target=increment) for _ in range(5)]

for t in threads:
    t.start()

for t in threads:
    t.join()

print(f'Final counter: {counter}')
```
