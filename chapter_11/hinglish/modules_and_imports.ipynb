# Modules, Packages aur Imports

Jaise aap apne kapde almirah me organize karke rakhte ho, waise hi code ko organize karne ke liye modules aur packages ka upyog hota hai.

Socho ki ek bade project pe kaam kar rahe ho. Sab kuch ek file me likhoge to bahut messy ho jayega. Isliye hum code ko different files me divide karte hai - har file ko **module** kehte hai.

## Module kya hai?

Module ek Python file hai jisme functions, classes, aur variables hote hai. Isko hum doosri files me import karke use kar sakte hai.

Jaise aapke paas ek dabba hai jisme different masale hai - haldi, mirchi, jeera. Zarurat ke hisab se nikal ke use karte ho!

### Khud ka module banana

Ek file banao `my_module.py`:

```python
# my_module.py
def greet(name):
    return f'Namaste {name}!'

def add(a, b):
    return a + b

PI = 3.14159
```

## Import ke different tarike

### 1. Poora module import karo

````python
import math
print(math.sqrt(16))
print(math.pi)
````

### 2. Specific cheez import karo

````python
from math import sqrt, pi
print(sqrt(25))
print(pi)
````

### 3. Sabkuch import karo (not recommended!)

````python
# from math import *
# print(sqrt(36))
````

### 4. Alias (nickname) dena

````python
import math as m
print(m.sqrt(49))

from datetime import datetime as dt
print(dt.now())
````

## Python ke Built-in Modules

### math module

````python
import math

print('Square root:', math.sqrt(64))
print('Power:', math.pow(2, 3))
print('Floor:', math.floor(3.7))
print('Ceil:', math.ceil(3.2))
````

### random module

````python
import random

print('Random number:', random.random())  # 0 to 1
print('Random int:', random.randint(1, 10))  # 1 to 10

fruits = ['aam', 'kela', 'seb']
print('Random choice:', random.choice(fruits))
````

### datetime module

````python
from datetime import datetime, date, timedelta

# Current date and time
print('Aaj:', datetime.now())
print('Date:', date.today())

# Future date
tomorrow = date.today() + timedelta(days=1)
print('Kal:', tomorrow)
````

### os module

````python
import os

print('Current directory:', os.getcwd())
print('Files:', os.listdir('.')[:5])  # Pehli 5 files
````

## Package kya hai?

Package modules ka collection hai. Ek folder jisme `__init__.py` file ho, wo package ban jata hai.

```
my_package/
    __init__.py
    module1.py
    module2.py
```

## Abhyaas

1. Calculator module banao (add, subtract, multiply, divide functions ke saath)
2. random module use karke dice roll simulator banao
3. datetime use karke apni age calculate karo
4. os module se ek folder me kitni files hai count karo

## Summary

1. **Module** - Python file with functions/classes
2. **Import** - module ko use karna
3. **Alias** - nickname dena
4. **Built-in modules** - math, random, datetime, os, etc.
5. **Package** - modules ka collection

Modules se code reusable aur organized rehta hai!
