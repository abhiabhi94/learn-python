# Websockets

Websockets real-time, two-way communication ke liye hai. Normal HTTP me client request karta hai aur server response deta hai. Par websockets me continuous connection rehta hai.

Jaise phone call - dono taraf se baat ho sakti hai simultaneously!

**Use cases:**
- Chat applications
- Live notifications
- Real-time data (stock prices, sports scores)
- Multiplayer games

## HTTP vs WebSocket

**HTTP:**
- Request-Response model
- Connection close hota hai
- New request ke liye new connection

**WebSocket:**
- Full-duplex communication
- Persistent connection
- Real-time updates
