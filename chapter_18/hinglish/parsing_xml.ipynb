# Parsing XML

XML (eXtensible Markup Language) data store aur transport karne ke liye use hota hai. Ye human-readable aur machine-readable hai.

Jaise HTML web pages ke liye hai, XML data ke liye hai!

**XML example:**
```xml
<student>
    <name>Ram</name>
    <age>18</age>
    <marks>85</marks>
</student>
```

## XML Structure

XML me:
- **Elements** - tags: `<name>Ram</name>`
- **Attributes** - `<student id="1">`
- **Nested elements** - elements ke andar elements
- **Root element** - sabse bahar wala element

## ElementTree Module

````python
import xml.etree.ElementTree as ET
````

## XML File padhna

````python
# Sample XML file banate hai
xml_data = '''
<students>
    <student id="1">
        <name>Ram</name>
        <age>18</age>
        <marks>85</marks>
    </student>
    <student id="2">
        <name>Priya</name>
        <age>17</age>
        <marks>92</marks>
    </student>
</students>
'''

# File me save karo
with open('students.xml', 'w') as f:
    f.write(xml_data)
````

### Parse karna

````python
# Parse XML file
tree = ET.parse('students.xml')
root = tree.getroot()

print('Root tag:', root.tag)
print('Root attributes:', root.attrib)
````

### Elements access karna

````python
# Sab students
for student in root.findall('student'):
    name = student.find('name').text
    age = student.find('age').text
    marks = student.find('marks').text
    student_id = student.get('id')
    
    print(f'ID: {student_id}, Name: {name}, Age: {age}, Marks: {marks}')
````

## XML String se Parse karna

````python
xml_string = '<book><title>Gitanjali</title><author>Tagore</author></book>'

root = ET.fromstring(xml_string)
print('Title:', root.find('title').text)
print('Author:', root.find('author').text)
````

## XML banana

````python
# Root element
root = ET.Element('library')

# Book element
book1 = ET.SubElement(root, 'book')
book1.set('id', '1')

title = ET.SubElement(book1, 'title')
title.text = 'Python Programming'

author = ET.SubElement(book1, 'author')
author.text = 'John Doe'

# Tree banao aur save karo
tree = ET.ElementTree(root)
tree.write('library.xml', encoding='utf-8', xml_declaration=True)
````

## XML modify karna

````python
# Load karo
tree = ET.parse('students.xml')
root = tree.getroot()

# Pehle student ke marks change karo
first_student = root.find('student')
marks = first_student.find('marks')
marks.text = '95'

# Save karo
tree.write('students_modified.xml')
````

## XPath-like Queries

````python
tree = ET.parse('students.xml')
root = tree.getroot()

# Find with specific attribute
student = root.find("./student[@id='2']")
if student is not None:
    print('Found:', student.find('name').text)

# All names
for name in root.findall('.//name'):
    print(name.text)
````

## Pretty Print

````python
# XML ko readable format me print karna
import xml.dom.minidom as minidom

def prettify(elem):
    rough_string = ET.tostring(elem, encoding='utf-8')
    reparsed = minidom.parseString(rough_string)
    return reparsed.toprettyxml(indent='  ')

# Use karo
tree = ET.parse('students.xml')
root = tree.getroot()
print(prettify(root))
````

## Practical Example - RSS Feed

````python
# RSS feed parse karna (simplified example)
rss_data = '''
<rss>
    <channel>
        <item>
            <title>News 1</title>
            <link>http://example.com/1</link>
        </item>
        <item>
            <title>News 2</title>
            <link>http://example.com/2</link>
        </item>
    </channel>
</rss>
'''

root = ET.fromstring(rss_data)
for item in root.findall('.//item'):
    title = item.find('title').text
    link = item.find('link').text
    print(f'{title}: {link}')
````

## XML vs JSON

**XML:**
- Verbose (zyada text)
- Schema validation
- Better for documents

**JSON:**
- Lightweight
- JavaScript friendly
- Better for APIs
