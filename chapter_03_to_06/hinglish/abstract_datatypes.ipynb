# Lists (Suchiyan)

Lists ek ordered collection (kramabaddh sangrah) hai jisme hum cheezein store kar sakte hai. Isko aap ek tarah ka container samajh sakte ho jisme aap items rakh sakte ho.

Example ke liye:
- Grocery list (kirana ki suchi)
- Ek class me students ki list
- Apne pasandida songs ki list

### List banana (Creating a list)

Square brackets `[]` ka upyog karke hum list bana sakte hai. Items ko comma (`,`) se alag karte hai.

```python
a_list = [1, 11, 111]
a_list
```

```python
a_list = ['Item 1', 'Item 2', 'Item 3']
```

```python
a_list
```

#### Prakar (Type)

`type()` function se hum kisi bhi cheez ka prakar (type) pata kar sakte hai.

```python
type(a_list)
```

#### Lambai (Length)

`len()` function se hum list me kitne elements hai, yeh pata kar sakte hai.

```python
len(a_list)
```

### Sthaan (Positions)

List me har element ka ek position hota hai jisko index kehte hai.

#### Shunya se shuru hone wala index (Zero based index)

Python me indexing 0 se shuru hoti hai. Matlab pehla element index 0 par hai, dusra index 1 par, aur aise hi aage.

```python
a_list = [1, 11, 111]
```

```python
a_list[0]
```

```python
a_list[1]
```

```python
a_list[2]
```

#### Nakaaratmak index (Negative index)

Python me hum negative index bhi use kar sakte hai. Negative index se hum list ke end se elements access kar sakte hai.

- `-1` matlab last element
- `-2` matlab last se pehla element
- Aur aise hi peeche ki taraf

```python
a_list[-1]
```

```python
a_list[-2]
```

```python
a_list[-3]
```

### Slicing (Tukde karna)

Slicing se hum list ka ek hissa (part) nikal sakte hai. Format hai: `list[start:stop:step]`

- `start`: Kaha se shuru karna hai
- `stop`: Kaha tak jana hai (yeh element include nahi hoga)
- `step`: Kitne-kitne elements skip karne hai

```python
a_list = [1, 11, 111, 1111, 11111]
```

```python
a_list[1:3]
```

```python
a_list[:3]
```

```python
a_list[2:]
```

```python
a_list[:]
```

### Elements jodna (Adding elements)

List me hum naye elements jod sakte hai.

#### `append()` method

`append()` se hum list ke end me ek element jod sakte hai.

```python
a_list = [1, 11, 111]
```

```python
a_list.append(1111)
```

```python
a_list
```

#### `extend()` method

`extend()` se hum ek list me dusri list ke saare elements jod sakte hai.

```python
a_list.extend([11111, 111111])
```

```python
a_list
```

#### `insert()` method

`insert()` se hum kisi particular position par element jod sakte hai.

Format: `list.insert(index, element)`

```python
a_list = [1, 11, 111]
```

```python
a_list.insert(1, 'new item')
```

```python
a_list
```

### Elements dhundhna (Searching elements)

List me elements dhundhne ke kai tarike hai.

#### `index()` method

`index()` se hum kisi element ka index pata kar sakte hai.

```python
a_list = [1, 11, 111, 1111]
```

```python
a_list.index(111)
```

#### `count()` method

`count()` se hum pata kar sakte hai ki koi element list me kitni baar aaya hai.

```python
a_list = [1, 11, 111, 11, 1111]
```

```python
a_list.count(11)
```

#### `in` operator ka upyog

`in` operator se hum check kar sakte hai ki koi element list me hai ya nahi.

```python
111 in a_list
```

```python
999 in a_list
```

### Elements hatana (Removing elements)

List se elements hatane ke kai tarike hai.

#### `pop()` method

`pop()` se hum kisi particular index par se element hata sakte hai. Default me yeh last element ko remove karta hai.

```python
a_list = [1, 11, 111, 1111]
```

```python
a_list.pop()
```

```python
a_list
```

```python
a_list.pop(1)
```

```python
a_list
```

#### `remove()` method

`remove()` se hum element ka naam dekar use hata sakte hai.

```python
a_list = [1, 11, 111, 1111]
```

```python
a_list.remove(111)
```

```python
a_list
```

# Tuples

Tuples bhi lists ki tarah hai, lekin ek important difference hai - **Tuples ko change nahi kar sakte** (immutable hai).

Tuples ko parentheses `()` se banate hai.

### Tuple banana (Creating a tuple)

```python
a_tuple = (11, 121, 122)
```

```python
a_tuple
```

```python
type(a_tuple)
```

### Tuple me indexing

Tuples me bhi indexing lists ki tarah hi kaam karti hai.

```python
a_tuple[0]
```

```python
a_tuple[-1]
```

### Lists aur Tuples me antar (Difference)

- Lists ko `[]` se banate hai, Tuples ko `()` se
- Lists **mutable** hai (badal sakte hai), Tuples **immutable** hai (nahi badal sakte)

```python
a_list = [1, 11, 111]
```

```python
a_list[0] = 999
```

```python
a_list
```

Lekin tuple me aisa nahi kar sakte:

```python
a_tuple = (11, 121, 122)
```

# Sets (Samuchay)

Sets ek unordered collection hai jisme **duplicate values nahi** hoti. Agar aapko sirf unique values chahiye, toh sets use kar sakte ho.

Sets ko curly braces `{}` se banate hai.

### Set banana (Creating a set)

```python
a_set = {1, 2, 3, 4, 5}
```

```python
a_set
```

```python
type(a_set)
```

### Duplicates automatically hat jate hai

Agar aap duplicate values daaloge, toh set automatically ek hi rakhega.

```python
a_set = {1, 2, 2, 3, 3, 3, 4, 4, 4, 4}
```

```python
a_set
```

### Set me elements jodna

`add()` method se ek element jod sakte hai.

```python
a_set = {1, 2, 3}
```

```python
a_set.add(4)
```

```python
a_set
```

### Set se elements hatana

`remove()` method se element hata sakte hai.

```python
a_set.remove(2)
```

```python
a_set
```

# Dictionaries (Shabdkosh)

Dictionaries ek bahut useful data structure hai. Isme hum **key-value pairs** store karte hai.

Jaise real dictionary me ek word (key) ka meaning (value) hota hai, waise hi Python dictionary me keys ke corresponding values hoti hai.

### Dictionary banana (Creating a dictionary)

Dictionaries ko curly braces `{}` se banate hai, lekin isme `key: value` pairs hote hai.

```python
info = {'name': 'Ram', 'age': 25, 'city': 'Delhi'}
```

```python
info
```

```python
type(info)
```

### Values access karna

Key ka naam dekar hum value access kar sakte hai.

```python
info['name']
```

```python
info['age']
```

### Naye key-value pairs jodna

```python
info['phone'] = '9876543210'
```

```python
info
```

### Values update karna

```python
info['age'] = 26
```

```python
info
```

### Keys aur values nikalna

```python
info.keys()
```

```python
info.values()
```

### Dictionary se items hatana

```python
info.pop('phone')
```

```python
info
```

## Tulna (Comparison)

Chaliye ab in sab data types ki tulna karte hai:

| Feature | List | Tuple | Set | Dictionary |
| ------- | ---- | ----- | --- | ---------- |
| Symbol | `[]` | `()` | `{}` | `{}` with key:value |
| Ordered | Haan | Haan | Nahi | Nahi (3.7+ me order maintain hota hai) |
| Mutable | Haan | Nahi | Haan | Haan |
| Duplicates | Allowed | Allowed | Not allowed | Keys unique honi chahiye |
| Indexing | Haan | Haan | Nahi | Keys se access |
| Use case | Jab order important ho | Jab data change nahi karna | Jab unique values chahiye | Jab key-value pairs chahiye |
