# Regular Expressions (Regex)

Regular expressions (regex) ek powerful tool hai patterns search karne ke liye. Jaise aap phone directory me sabhi 9 se start hone wale numbers dhoondhna chahte ho.

Regex se hum:
- Email validate kar sakte hai
- Phone numbers extract kar sakte hai
- Patterns search kar sakte hai
- Text replace kar sakte hai

## re module

Python me regex ke liye `re` module use karte hai.

```python
import re
```

## Basic Methods

### search() - pehla match dhoondhna

```python
text = 'Mera phone number 9876543210 hai'

# Digit search karna
result = re.search(r'\d+', text)
if result:
    print('Mila:', result.group())
```

### findall() - sab matches dhoondhna

```python
text = 'Ram ki age 25 hai aur Shyam ki 30'

# Sab numbers dhoondhna
numbers = re.findall(r'\d+', text)
print('Numbers:', numbers)
```

### match() - string ke start se match karna

```python
text = '9876543210'

if re.match(r'\d{10}', text):
    print('Valid 10-digit number')
```

## Common Patterns

### Basic patterns:
- `\d` - digit (0-9)
- `\w` - word character (a-z, A-Z, 0-9, _)
- `\s` - whitespace
- `.` - any character
- `^` - start of string
- `$` - end of string

```python
# Email validation
email = 'ram@gmail.com'
if re.match(r'^[\w.-]+@[\w.-]+\.\w+$', email):
    print('Valid email')
```

### Quantifiers:
- `*` - 0 ya zyada baar
- `+` - 1 ya zyada baar
- `?` - 0 ya 1 baar
- `{n}` - exactly n baar
- `{n,m}` - n se m baar

```python
# Phone number pattern
phone = '9876543210'
if re.match(r'\d{10}', phone):
    print('Valid phone number')
```

## sub() - Replace karna

```python
text = 'Mera number 9876543210 hai'

# Numbers ko XXX se replace karo
new_text = re.sub(r'\d', 'X', text)
print(new_text)
```

## split() - Split karna

```python
text = 'ram,shyam;mohan:priya'

# Multiple delimiters se split
names = re.split(r'[,;:]', text)
print(names)
```

## Groups

Parentheses () se groups banate hai.

```python
text = 'Mera email ram@gmail.com hai'

# Groups capture karna
match = re.search(r'([\w.-]+)@([\w.-]+)', text)
if match:
    print('Username:', match.group(1))
    print('Domain:', match.group(2))
```

## Practical Examples

### 1. Extract all emails

```python
text = 'Contact us at info@company.com or support@company.co.in'

emails = re.findall(r'[\w.-]+@[\w.-]+\.\w+', text)
print('Emails found:', emails)
```

### 2. Validate Indian mobile number

```python
def validate_mobile(number):
    # 10 digits, start with 6-9
    pattern = r'^[6-9]\d{9}$'
    return bool(re.match(pattern, number))

print(validate_mobile('9876543210'))  # True
print(validate_mobile('1234567890'))  # False
```

### 3. Extract dates

```python
text = 'Meeting on 15-08-2023 and 20-09-2023'

dates = re.findall(r'\d{2}-\d{2}-\d{4}', text)
print('Dates:', dates)
```
